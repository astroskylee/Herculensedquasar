# Projected M/L gradient

```python
import os
os.environ['HDF5_USE_FILE_LOCKING'] = 'FALSE'

from pathlib import Path

import arviz as az
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams.update({
    'figure.dpi': 140,
    'savefig.dpi': 300,
    'font.size': 11,
})
```

```python
NC_PATH = '/mnt/lustre/tianli/quasar_hmc/WFI2033_ss=2_inferh0_step6_imaging_only_infer_LCDM_20260422_15/WFI2033_all_ss=2_inferh0_step6_imaging_only_infer_LCDM.nc'

idata = az.from_netcdf(NC_PATH)
post = idata.posterior
num_chains = int(post.sizes['chain'])

print('NC_PATH =', NC_PATH)
div = idata.sample_stats['diverging'].sum(dim='draw').values
print('divergences per chain:')
print(div)
```

```python
suffix_epl = '_ss=2_full_light_multimass'
run_tag = '20260417_14'

SCRIPT_DIR = Path.cwd()
RESULT_DIR = SCRIPT_DIR / 'result' / f'result{suffix_epl}_{run_tag}'
FIXED_FIVE_GAUSS_PATH = RESULT_DIR / f'fixed_five_gaussians_corrected_psf{suffix_epl}.npz'

fixed_five_gaussians = np.load(FIXED_FIVE_GAUSS_PATH)
light_amp = fixed_five_gaussians['amp']
sigma = fixed_five_gaussians['sigma']
light_weight = light_amp / light_amp.sum()

m2l = post['m2l_ratio'].values.reshape(-1)
m2l_slope = post['m2l_ratio_slope'].values.reshape(-1)

m2l_med = np.median(m2l)
m2l_slope_med = np.median(m2l_slope)

print('FIXED_FIVE_GAUSS_PATH =', FIXED_FIVE_GAUSS_PATH)
print('median m2l_ratio =', m2l_med)
print('median m2l_ratio_slope =', m2l_slope_med)
```

```python
def projected_mge_profile(R, amp, sigma):
    kernel = np.exp(-0.5 * (R[:, None] / sigma[None, :])**2) / (2.0 * np.pi * sigma[None, :]**2)
    return (amp[None, :] * kernel).sum(axis=1)


R = np.linspace(0.0, 1.0, 300)

I_light = projected_mge_profile(R, light_weight, sigma)

ml_gauss_med = m2l_med * sigma**m2l_slope_med
mass_amp_med = light_weight * ml_gauss_med
Sigma_star_med = projected_mge_profile(R, mass_amp_med, sigma)
ML_med = Sigma_star_med / I_light
```

```python
rng = np.random.default_rng(42)
sample_idx = rng.choice(m2l.size, size=2000, replace=False)

ML_samples = []
for i in sample_idx:
    ml_gauss_i = m2l[i] * sigma**m2l_slope[i]
    mass_amp_i = light_weight * ml_gauss_i
    Sigma_star_i = projected_mge_profile(R, mass_amp_i, sigma)
    ML_samples.append(Sigma_star_i / I_light)

ML_samples = np.stack(ML_samples)
ML_16, ML_50, ML_84 = np.percentile(ML_samples, [16, 50, 84], axis=0)
```

```python
fig, ax = plt.subplots(figsize=(4.8, 3.6))

ax.fill_between(R, ML_16, ML_84, color='0.82', linewidth=0, label='16-84%')
ax.plot(R, ML_50, color='black', linewidth=1.8, label='posterior median')
ax.plot(R, ML_med, color='tab:red', linewidth=1.0, linestyle='--', label='median parameters')

ax.set_xlim(0.0, 1.0)
ax.set_xlabel('Projected radius R [arcsec]')
ax.set_ylabel('Projected stellar M/L')
ax.legend(frameon=False)
fig.tight_layout()

fig.savefig(SCRIPT_DIR / 'result' / 'projected_ml_gradient_infer_lcdm_20260422_15.png', bbox_inches='tight')
```
