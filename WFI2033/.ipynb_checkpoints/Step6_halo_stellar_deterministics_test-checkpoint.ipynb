# Step 6 halo and stellar-mass derived posterior test

Read the two derived Step 6 netCDF files, draw 100 posterior samples from each, and compare the derived halo and stellar quantities with a scatter-style corner plot.


```python
import os
os.environ.setdefault("HDF5_USE_FILE_LOCKING", "FALSE")

from pathlib import Path

import arviz as az
import corner
import matplotlib.pyplot as plt
import numpy as np
import xarray as xr
```

```python
PANTHEON_SH0ES_DERIVED_NC = Path(
    "/mnt/lustre/tianli/quasar_hmc/"
    "WFI2033_ss=2_fullconcen_light_multimass_step6_pantheonsh0es_reuse_redwarmup_20260507_14/"
    "WFI2033_all_ss=2_fullconcen_light_multimass_step6_pantheonsh0es_reuse_redwarmup_halo_stellar_deterministics.nc"
)

DESI_PLANCK_DERIVED_NC = Path(
    "/mnt/lustre/tianli/quasar_hmc/"
    "WFI2033_ss=2_fullconcen_light_multimass_step6_desi_planck_20260504_16/"
    "WFI2033_all_ss=2_fullconcen_light_multimass_step6_desi_planck_halo_stellar_deterministics.nc"
)

VARS = [
    "M200_halo_1E12",
    "c200_halo",
    "R200_halo_kpc",
    "total_stellar_mass_1E11",
    "m2l_ratio",
    "m2l_ratio_slope",
]

LABELS = [
    r"$M_{200}\,[10^{12}M_\odot]$",
    r"$c_{200}$",
    r"$R_{200}\,[\mathrm{kpc}]$",
    r"$M_\star\,[10^{11}M_\odot]$",
    r"$M/L$",
    r"$\nabla_{M/L}$",
]

N_SAMPLE = 100
RNG_SEED = 20260512

print("PANTHEON_SH0ES_DERIVED_NC =", PANTHEON_SH0ES_DERIVED_NC)
print("DESI_PLANCK_DERIVED_NC =", DESI_PLANCK_DERIVED_NC)
```

```python
def load_flat_sample(nc_path, var_names, n_sample=100, seed=0):
    post = xr.open_dataset(nc_path, group="posterior", engine="h5netcdf")[var_names].load()
    flat = post.stack(sample=("chain", "draw")).transpose("sample", ...)
    rng = np.random.default_rng(seed)
    sample_index = rng.choice(flat.sizes["sample"], size=n_sample, replace=False)
    flat_sample = flat.isel(sample=sample_index)
    data = np.column_stack([np.asarray(flat_sample[name].values, dtype=float).reshape(n_sample, -1)[:, 0] for name in var_names])
    return data, flat_sample

pantheon_sample, pantheon_xr = load_flat_sample(PANTHEON_SH0ES_DERIVED_NC, VARS, N_SAMPLE, RNG_SEED + 1)
desi_sample, desi_xr = load_flat_sample(DESI_PLANCK_DERIVED_NC, VARS, N_SAMPLE, RNG_SEED + 2)

print("Pantheon+SH0ES sample shape:", pantheon_sample.shape)
print("DESI+Planck sample shape:", desi_sample.shape)
print("Pantheon+SH0ES medians:", dict(zip(VARS, np.nanmedian(pantheon_sample, axis=0))))
print("DESI+Planck medians:", dict(zip(VARS, np.nanmedian(desi_sample, axis=0))))
```

```python
fig = corner.corner(
    desi_sample,
    labels=LABELS,
    color="#356ab7",
    plot_datapoints=True,
    plot_density=False,
    plot_contours=False,
    fill_contours=False,
    hist_kwargs={"density": True, "histtype": "step", "linewidth": 1.8},
    data_kwargs={"alpha": 0.75, "ms": 3.5},
    label_kwargs={"fontsize": 15},
)

corner.corner(
    pantheon_sample,
    labels=LABELS,
    fig=fig,
    color="#c44e52",
    plot_datapoints=True,
    plot_density=False,
    plot_contours=False,
    fill_contours=False,
    hist_kwargs={"density": True, "histtype": "step", "linewidth": 1.8},
    data_kwargs={"alpha": 0.75, "ms": 3.5},
)

fig.legend(
    handles=[
        plt.Line2D([], [], color="#356ab7", marker=".", linestyle="None", markersize=8, label="DESI+Planck Cosmology"),
        plt.Line2D([], [], color="#c44e52", marker=".", linestyle="None", markersize=8, label="Pantheon+SH0ES Cosmology"),
    ],
    loc="upper left",
    bbox_to_anchor=(0.70, 0.96),
    frameon=False,
    fontsize=14,
)

for ax in fig.get_axes():
    ax.tick_params(axis="both", which="major", labelsize=11, direction="in")
    ax.tick_params(axis="both", which="minor", direction="in")

plt.savefig("./Paper/step6_derived_halo_stellar_100sample_corner_scatter.pdf", dpi=300, bbox_inches="tight")
plt.show()
```
