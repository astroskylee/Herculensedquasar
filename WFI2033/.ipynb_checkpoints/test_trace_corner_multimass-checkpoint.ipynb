# Test Trace And Corner

Minimal notebook for reading one HMC `.nc` file and plotting trace/corner diagnostics.

```python
import os
os.environ['HDF5_USE_FILE_LOCKING'] = 'FALSE'

from pathlib import Path

import arviz as az
import corner
import matplotlib.pyplot as plt
import numpy as np
```

```python
NC_PATH = Path('/mnt/lustre/tianli/quasar_hmc/WFI2033_ss=2_full_light_multimass_20260401_00/WFI2033_2_ss=2_full_light_multimass.nc')

inf_data_pixel = az.from_netcdf(str(NC_PATH))
post = inf_data_pixel.posterior

print('NC_PATH =', NC_PATH)
print('posterior dims =', post.dims)
print('posterior sizes =', dict(post.sizes))
print('data vars =', list(post.data_vars))
```

```python
vars_mass = [
    'theta_E_1', 'theta_E_g1', 'theta_E_g2', 'theta_E_g3', 'theta_E_g7',
    'gamma_1', 'e_1', 'center_1', 'gamma_sheer_1',
]
vars_power = ['n_source_grid', 'rho_source_grid', 'sigma_source_grid']
vars_point_source = ['ra_ps', 'dec_ps', 'log10_amp_ps']

available_vars = [
    name for name in (vars_mass + vars_power + vars_point_source)
    if name in post.data_vars
]

print('available vars for diagnostics =', available_vars)
```

```python
plt.rcParams['figure.constrained_layout.use'] = True
_ = az.plot_trace(
    inf_data_pixel,
    var_names=available_vars,
    figsize=(14, 18),
    divergences=True,
)
plt.show()
```

```python
corner_vars = [name for name in vars_mass if name in post.data_vars]

fig_corner = None
num_chains = int(post.sizes['chain'])
for i in range(num_chains):
    fig_corner = corner.corner(
        post.isel(chain=i),
        var_names=corner_vars,
        color=f'C{i}',
        fig=fig_corner,
    )
plt.show()
```

```python
summary_vars = [name for name in vars_mass + vars_power if name in post.data_vars]
az.summary(inf_data_pixel, var_names=summary_vars)
```
